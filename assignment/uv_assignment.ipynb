## Task 3: Create a package and CLI entrypoint
Create a package named `assignment_uv_project` with a `cli.py` and an entry `assignment-hello`.

Requirements:
- Package directory with `__init__.py` and `cli.py`
- `[project.scripts]` includes `assignment-hello = "assignment_uv_project.cli:main"`
- Implement `main()` that prints `Hello, <name>` (default `world`)
- Demonstrate running: `uv run assignment-hello` and `uv run assignment-hello YourName`


````python
%%writefile assignment_uv_project/__init__.py
__version__ = '0.1.0'
````

````python
%%writefile assignment_uv_project/cli.py
from __future__ import annotations
import sys

def main() -> int:
    name = sys.argv[1] if len(sys.argv) > 1 else "world"
    print(f"Hello, {name} 👋 (from assignment_uv_project)")
    return 0

if __name__ == "__main__":
    raise SystemExit(main())
````

````python
# Ensure tests directory exists before writing files
mkdir -p tests
````

````python
%%writefile tests/test_cli.py
from assignment_uv_project.cli import main


def test_main_default(capsys):
    assert main() == 0
    out = capsys.readouterr().out
    assert "Hello, world" in out
````

````python
%%writefile -a pyproject.toml
[project.scripts]
assignment-hello = "assignment_uv_project.cli:main"
````
